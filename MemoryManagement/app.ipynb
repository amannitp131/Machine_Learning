# Tracemalloc in python

```python
import tracemalloc

# Start tracking memory
tracemalloc.start()

# Take a snapshot before code
snapshot1 = tracemalloc.take_snapshot()

# Some sample code that uses memory
x = [i ** 2 for i in range(100000)] 

# Take a snapshot after code
snapshot2 = tracemalloc.take_snapshot()

# Compare both snapshots
top_stats = snapshot2.compare_to(snapshot1, 'lineno')

# Print top 5 memory usage differences
for stat in top_stats[:5]:
    print(stat)
```

# Circular reference

```python
import gc

class MyObject:
    def __init__(self, name): 
        self.name = name
        print(f"Object {self.name} created")

    def __del__(self):
        print(f"Object {self.name} deleted")


obj1 = MyObject("obj1")
obj2 = MyObject("obj2")

# Create circular reference
obj1.ref = obj2
obj2.ref = obj1

# Delete references
del obj1
del obj2

# Force garbage collection
gc.collect()  # 🔁 Helps clean up circular references
```
