## 🛡️ 2. HTTP Headers & 403 Error Handling

### 🎯 Learning Objectives:
- **Understand** HTTP 403 Forbidden errors and their causes
- **Learn** how User-Agent headers help bypass basic restrictions
- **Master** professional header configuration for web scraping

### 📋 Key Concepts:
- **403 Forbidden**: Server denies access to automated requests
- **User-Agent**: Browser identification string in HTTP headers
- **Header Spoofing**: Making requests appear to come from real browsers
- **Anti-Bot Protection**: Common website defense mechanisms

### 💡 When You Get 403 Errors:
Websites often block requests that don't include proper browser headers. The solution is to include a realistic User-Agent string that mimics a real browser.

### 🔧 Professional Header Configuration:
```python
headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 6.3; Win 64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/80.0.3987.162 Safari/537.36'
}
response = requests.get('url', headers=headers)
webpage_content = response.text
```

### ⚠️ **Important Notes:**
- **Respect robots.txt**: Always check website's scraping policy
- **Rate Limiting**: Don't overwhelm servers with rapid requests
- **Legal Compliance**: Ensure scraping complies with terms of service
- **Ethical Scraping**: Use scraped data responsibly

### 🛠️ **Best Practices:**
- ✅ **Always include User-Agent** headers
- ✅ **Add delays** between requests
- ✅ **Handle errors** gracefully
- ✅ **Monitor response codes** (200, 403, 404, etc.)

## 📥 4. Initial Web Request to Target Site

### 🎯 Learning Objectives:
- **Execute** HTTP GET request to target website
- **Retrieve** HTML content for parsing
- **Handle** potential connection issues
- **Prepare** content for BeautifulSoup processing

### 📋 Target Website:
- **Site**: AmbitionBox (Company reviews and information)
- **URL**: https://www.ambitionbox.com/list-of-companies
- **Purpose**: Extract company data including ratings, reviews, and details
- **Method**: GET request to retrieve HTML content

### 🔧 Implementation:

````python
# Make initial web request to target website
print("🌐 Making web request to AmbitionBox...")

try:
    # Target URL for company listings
    target_url = 'https://www.ambitionbox.com/list-of-companies?page=1'
    
    # Make the web request
    response = requests.get(target_url)
    
    # Extract HTML content
    webpage = response.text
    
    print(f"✅ Web request successful!")
    print(f"📊 Status Code: {response.status_code}")
    print(f"📄 Content Length: {len(webpage):,} characters")
    print(f"🌐 Final URL: {response.url}")
    print(f"📋 Content Type: {response.headers.get('content-type', 'Unknown')}")
    
    # Quick content validation
    if len(webpage) > 1000:
        print("✅ Substantial content retrieved - ready for parsing")
    else:
        print("⚠️ Limited content retrieved - check for access restrictions")
        
except requests.exceptions.RequestException as e:
    print(f"❌ Request failed: {e}")
    print("💡 Try using headers if you encounter 403 errors")
except Exception as e:
    print(f"❌ Unexpected error: {e}")
````

## 🍲 5. HTML Parsing with BeautifulSoup

### 🎯 Learning Objectives:
- **Parse** raw HTML content into navigable structure
- **Understand** BeautifulSoup's parsing capabilities
- **Prepare** HTML for element extraction and manipulation
- **Choose** appropriate parser for optimal performance

### 📋 Key Concepts:
- **HTML Parser**: Engine that converts HTML text into object structure
- **lxml Parser**: Fast C-based parser (recommended for performance)
- **Soup Object**: BeautifulSoup's representation of HTML document
- **DOM Navigation**: Accessing elements through hierarchical structure

### 💡 Parser Options:
- **lxml**: Fastest, most feature-complete (recommended)
- **html.parser**: Built-in Python parser (slower but no dependencies)
- **xml**: For XML documents specifically
- **html5lib**: Most lenient parser (handles malformed HTML well)

### 🔧 Implementation:

````python
# Parse HTML content with BeautifulSoup
print("🍲 Parsing HTML content with BeautifulSoup...")

try:
    # Create BeautifulSoup object with lxml parser for best performance
    soup = BeautifulSoup(webpage, 'lxml')
    
    print(f"✅ HTML parsing completed successfully!")
    print(f"📄 Parser Used: lxml (fastest and most feature-complete)")
    print(f"🏷️ Document Title: {soup.title.string if soup.title else 'No title found'}")
    
    # Basic structure analysis
    all_tags = [tag.name for tag in soup.find_all()]
    unique_tags = set(all_tags)
    
    print(f"📊 HTML Structure Analysis:")
    print(f"   Total Elements: {len(all_tags):,}")
    print(f"   Unique Tag Types: {len(unique_tags)}")
    print(f"   Common Tags: {sorted(list(unique_tags))[:10]}")
    
    # Check for common scraping targets
    forms = len(soup.find_all('form'))
    divs = len(soup.find_all('div'))
    links = len(soup.find_all('a'))
    
    print(f"\n🔍 Content Overview:")
    print(f"   Forms: {forms}")
    print(f"   Div containers: {divs:,}")
    print(f"   Links: {links:,}")
    
    print("✅ Soup object ready for element extraction!")
    
except Exception as e:
    print(f"❌ Parsing error: {e}")
    print("💡 Try different parser: 'html.parser' or 'html5lib'")
````

## 🏢 8. Company Names Extraction

### 🎯 Learning Objectives:
- **Extract** company names from h2 elements
- **Process** multiple elements using loops
- **Clean** extracted text data
- **Understand** element iteration patterns

### 📋 Key Concepts:
- **h2 Elements**: Secondary headings typically containing company names
- **Loop Processing**: Iterating through multiple elements
- **Text Cleaning**: Using strip() to remove whitespace
- **Data Collection**: Building lists from scraped content

### 💡 Why h2 Elements?
Company names are often displayed in h2 headings for SEO and accessibility. This makes them easy targets for extraction.

### 🔧 Implementation:

````python
# Extract company names from h2 elements
print("🏢 Extracting company names from h2 elements...")

try:
    # Find all h2 elements (typically contain company names)
    h2_elements = soup.find_all('h2')
    
    print(f"📊 Found {len(h2_elements)} h2 elements")
    print(f"🏢 Extracting company names:")
    print("-" * 50)
    
    company_names = []
    for i, element in enumerate(h2_elements, 1):
        company_name = element.text.strip()
        company_names.append(company_name)
        print(f"{i:2d}. {company_name}")
    
    print("-" * 50)
    print(f"✅ Extracted {len(company_names)} company names")
    
    # Data quality check
    if company_names:
        avg_length = sum(len(name) for name in company_names) / len(company_names)
        print(f"📈 Average name length: {avg_length:.1f} characters")
        print(f"📏 Longest name: {max(company_names, key=len)} ({len(max(company_names, key=len))} chars)")
        print(f"📏 Shortest name: {min(company_names, key=len)} ({len(min(company_names, key=len))} chars)")
    
except Exception as e:
    print(f"❌ Error extracting company names: {e}")
````

## 📦 12. Container-Based Data Extraction

### 🎯 Learning Objectives:
- **Identify** main container elements holding company data
- **Understand** structured data organization in HTML
- **Implement** container-based scraping strategy
- **Extract** multiple data points from unified containers

### 📋 Key Concepts:
- **Container Elements**: div elements that group related information
- **Data Grouping**: HTML structure organizing company information
- **Wrapper Classes**: CSS classes identifying content containers
- **Structured Extraction**: Getting all data from organized sections

### 💡 Why Container-Based Extraction?
Instead of searching for individual elements scattered across the page, container-based extraction finds the main content blocks first, then extracts all relevant data from each container. This ensures data consistency and proper association.

### 🔧 Container Strategy:
1. **Find** main content containers
2. **Iterate** through each container  
3. **Extract** all relevant fields from each container
4. **Maintain** data relationships and consistency

---

````python
# Identify and extract main company content containers
print("📦 Identifying company content containers...")

try:
    # Find all div elements with 'company-content-wrapper' class
    company_containers = soup.find_all('div', class_='company-content-wrapper')
    
    print(f"✅ Found {len(company_containers)} company containers")
    
    if company_containers:
        print(f"📊 Container Analysis:")
        print(f"   Container class: 'company-content-wrapper'")
        print(f"   Total containers: {len(company_containers)}")
        print(f"   Element type: div")
        
        # Analyze first container structure
        if len(company_containers) > 0:
            first_container = company_containers[0]
            child_elements = first_container.find_all()
            
            print(f"\n🔍 Sample Container Structure:")
            print(f"   Child elements in first container: {len(child_elements)}")
            
            # Show element types in first container
            element_types = {}
            for element in child_elements:
                tag_class = element.get('class', ['no-class'])
                key = f"{element.name}.{'.'.join(tag_class)}"
                element_types[key] = element_types.get(key, 0) + 1
            
            print(f"   Element distribution:")
            for element_type, count in sorted(element_types.items())[:8]:
                print(f"     {element_type}: {count}")
                
        print(f"\n✅ Container identification successful!")
        print(f"🎯 Ready for comprehensive data extraction")
    else:
        print("❌ No company containers found")
        print("💡 Check if the class name 'company-content-wrapper' is correct")
        print("💡 Try inspecting the HTML for actual container classes")
        
except Exception as e:
    print(f"❌ Error identifying containers: {e}")
````

````python
# Verify container count for data validation
print(f"📊 Container Validation:")
print(f"   Total company containers found: {len(company_containers)}")
print(f"   Expected data points per container: 7 fields")
print(f"   Estimated total data points: {len(company_containers) * 7}")
print("✅ Container count confirmed - ready for extraction")
````

## 🎯 13. Comprehensive Data Extraction

### 🎯 Learning Objectives:
- **Extract** multiple data fields from each company container
- **Handle** missing data gracefully with error handling
- **Build** structured dataset from scraped information
- **Create** pandas DataFrame for analysis

### 📋 Data Fields to Extract:
1. **Company Name** (h2 element)
2. **Rating** (p.rating element)  
3. **Review Count** (a.review-count element)
4. **Company Type** (first p.infoEntity)
5. **Headquarters** (second p.infoEntity)
6. **Company Age** (third p.infoEntity)
7. **Employee Count** (fourth p.infoEntity)

### 💡 Extraction Strategy:
- **Container-based**: Extract all fields from each company container
- **Error Handling**: Use try-except blocks for missing elements
- **Data Structure**: Build lists for each field type
- **DataFrame Creation**: Convert lists to structured pandas DataFrame

### 🔧 Implementation:

````python
# Comprehensive data extraction from company containers
print("🎯 Starting comprehensive data extraction...")
print("=" * 60)

# Initialize lists to store extracted data
name = []
rating = []
reviews = []
ctype = []
hq = []
how_old = []
no_of_employee = []

print(f"📊 Processing {len(company_containers)} company containers...")

# Extract data from each company container
for i, container in enumerate(company_containers, 1):
    print(f"Processing company {i}/{len(company_containers)}", end="\r")
    
    try:
        # Extract company name from h2 element
        name_element = container.find('h2')
        name.append(name_element.text.strip() if name_element else "N/A")
        
        # Extract rating from p.rating element
        rating_element = container.find('p', class_='rating')
        rating.append(rating_element.text.strip() if rating_element else "N/A")
        
        # Extract review count from a.review-count element
        review_element = container.find('a', class_='review-count')
        reviews.append(review_element.text.strip() if review_element else "N/A")
        
        # Extract company information from infoEntity elements
        info_entities = container.find_all('p', class_='infoEntity')
        
        # Company type (first infoEntity)
        ctype.append(info_entities[0].text.strip() if len(info_entities) > 0 else "N/A")
        
        # Headquarters (second infoEntity)
        hq.append(info_entities[1].text.strip() if len(info_entities) > 1 else "N/A")
        
        # Company age (third infoEntity)
        how_old.append(info_entities[2].text.strip() if len(info_entities) > 2 else "N/A")
        
        # Number of employees (fourth infoEntity)
        no_of_employee.append(info_entities[3].text.strip() if len(info_entities) > 3 else "N/A")
        
    except Exception as e:
        print(f"\n⚠️ Error processing container {i}: {e}")
        # Add placeholder data for failed extraction
        name.append("ERROR")
        rating.append("ERROR")
        reviews.append("ERROR")
        ctype.append("ERROR")
        hq.append("ERROR")
        how_old.append("ERROR")
        no_of_employee.append("ERROR")

print(f"\n✅ Data extraction completed!")

# Create DataFrame from extracted data
df = pd.DataFrame({
    'name': name,
    'rating': rating,
    'reviews': reviews,
    'company_type': ctype,
    'Head_Quarters': hq,
    'Company_Age': how_old,
    'No_of_Employee': no_of_employee
})

print(f"\n📊 DataFrame Creation Summary:")
print(f"   Companies processed: {len(company_containers)}")
print(f"   DataFrame shape: {df.shape}")
print(f"   Columns: {df.columns.tolist()}")

# Data quality check
print(f"\n🔍 Data Quality Assessment:")
for col in df.columns:
    missing_count = (df[col] == "N/A").sum()
    error_count = (df[col] == "ERROR").sum()
    print(f"   {col}: {missing_count} missing, {error_count} errors")

print(f"\n✅ Single-page extraction complete - DataFrame ready!")
````
