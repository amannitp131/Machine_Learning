## 🌐 2. API Introduction & First Request

### 🎯 Learning Objectives:
- **Understand** TMDb API structure and endpoints
- **Learn** API authentication using API keys
- **Master** HTTP GET requests for data retrieval
- **Handle** API responses and status codes

### 📋 Key Concepts:
- **API Endpoint**: Specific URL for data access
- **API Key**: Authentication token for API access
- **Query Parameters**: URL parameters for filtering data
- **JSON Response**: Structured data format from API

### 💡 TMDb API Structure:
```
Base URL: https://api.themoviedb.org/3/
Endpoint: movie/top_rated
Parameters: api_key, language, page
```

### 🔧 API Parameters Explained:
- **api_key**: Your authentication token
- **language**: Response language (en-US for English)
- **page**: Page number for pagination (starts from 1)

---

````python
# Make first API request to The Movie Database (TMDb)
# Fetch top-rated movies from page 1

print("🌐 Making API request to TMDb...")

# API endpoint configuration
base_url = "https://api.themoviedb.org/3/movie/top_rated"
api_key = "8265bd1679663a7ea12ac168da84d2e8"
params = {
    "api_key": api_key,
    "language": "en-US", 
    "page": 1
}

# Make the API request
response = requests.get(base_url, params=params)

# Check if request was successful
if response.status_code == 200:
    data = response.json()
    print(f"📋 Response Keys: {list(data.keys())}")
    print(f"🎬 Movies in this page: {len(data.get('results', []))}")
    print(f"📄 Total Pages Available: {data.get('total_pages', 'N/A')}")
    print(f"🎭 Total Movies: {data.get('total_results', 'N/A')}")
else:
    print(f"❌ API request failed with status code: {response.status_code}")
````

## 🔄 6. Batch Data Collection Loop

### 🎯 Learning Objectives:
- **Implement** large-scale data collection using loops
- **Handle** API pagination across multiple pages
- **Manage** rate limiting and API best practices
- **Process** and accumulate data efficiently

### 📋 Key Concepts:
- **Pagination**: Accessing data across multiple pages
- **Loop Processing**: Iterating through page numbers
- **Data Concatenation**: Combining DataFrames efficiently
- **Progress Tracking**: Monitoring collection progress

### ⚠️ Important Considerations:
- **API Limits**: Respect rate limiting (requests per second/minute)
- **Error Handling**: Handle network issues gracefully
- **Memory Usage**: Monitor memory consumption during collection
- **Data Validation**: Ensure data consistency across pages

### 💡 Collection Strategy:
- **Pages 1-428**: Based on total_pages from initial request
- **Incremental Building**: Add data page by page
- **Progress Monitoring**: Track collection status

### 🚨 **Note**: This is a long-running operation that may take several minutes!

---

````python
# Batch data collection from multiple API pages
# WARNING: This will make 428 API requests - may take several minutes!

print("🔄 Starting batch data collection...")
print("⚠️  This will collect data from 428 pages - please be patient!")

# Collection configuration
total_pages = 428  # Based on API response
collected_pages = 0
failed_requests = 0
base_url = "https://api.themoviedb.org/3/movie/top_rated"
api_key = "8265bd1679663a7ea12ac168da84d2e8"

# Progress tracking
progress_interval = 50  # Show progress every 50 pages

try:
    for page_num in range(1, total_pages + 1):
        try:
            # Make API request for current page
            params = {
                "api_key": api_key,
                "language": "en-US",
                "page": page_num
            }
            response = requests.get(base_url, params=params)
            
            if response.status_code == 200:
                # Extract and process data
                movies_data = response.json()['results']
                selected_columns = ['id', 'title', 'overview', 'release_date', 'popularity', 'vote_average', 'vote_count']
                temp_df = pd.DataFrame(movies_data)[selected_columns]
                
                # Append to main DataFrame (using concat instead of deprecated append)
                df = pd.concat([df, temp_df], ignore_index=True)
                collected_pages += 1
                
                # Progress reporting
                if page_num % progress_interval == 0:
                    print(f"📊 Progress: {page_num}/{total_pages} pages collected ({collected_pages * 20} movies)")
                    
            else:
                print(f"❌ Failed to fetch page {page_num}: Status {response.status_code}")
                failed_requests += 1
                
        except Exception as e:
            print(f"❌ Error on page {page_num}: {e}")
            failed_requests += 1
            continue

    # Final summary
    print(f"\n🎉 Batch collection completed!")
    print(f"✅ Successfully collected: {collected_pages} pages")
    print(f"❌ Failed requests: {failed_requests}")
    print(f"📊 Final dataset shape: {df.shape}")
    print(f"🎬 Total movies collected: {len(df)}")
    print(f"💾 Memory usage: {df.memory_usage(deep=True).sum() / 1024 / 1024:.2f} MB")

except KeyboardInterrupt:
    print(f"\n⏹️ Collection interrupted by user")
    print(f"📊 Partial data collected: {df.shape}")
except Exception as e:
    print(f"\n❌ Critical error during collection: {e}")
    print(f"📊 Data collected so far: {df.shape}")
````

## 📊 7. Final Dataset Analysis

### 🎯 Learning Objectives:
- **Analyze** the complete collected dataset
- **Validate** data quality and completeness
- **Explore** statistical distributions and patterns
- **Assess** collection success and data insights

### 🔧 Analysis Components:
- **Dataset Overview**: Shape, size, and structure
- **Data Quality**: Missing values and consistency
- **Statistical Summary**: Distributions and ranges
- **Sample Exploration**: Representative data examples

---

````python
# Comprehensive analysis of the final collected dataset
print("📊 Final Dataset Analysis")
print("=" * 60)

try:
    # Basic dataset information
    print(f"🎬 Complete Movie Database Collection Summary:")
    print(f"   📊 Dataset Shape: {df.shape}")
    print(f"   📋 Columns: {df.columns.tolist()}")
    print(f"   💾 Memory Usage: {df.memory_usage(deep=True).sum() / 1024 / 1024:.2f} MB")
    print(f"   🎭 Total Movies: {len(df):,}")
    
    # Data quality assessment
    print(f"\n🔍 Data Quality Assessment:")
    missing_data = df.isnull().sum()
    total_missing = missing_data.sum()
    
    if total_missing > 0:
        print(f"   ⚠️ Total missing values: {total_missing}")
        for col, missing in missing_data[missing_data > 0].items():
            percentage = (missing / len(df)) * 100
            print(f"     {col}: {missing:,} ({percentage:.2f}%)")
    else:
        print("   ✅ No missing values detected!")
    
    # Duplicate check
    duplicates = df.duplicated().sum()
    print(f"   🔁 Duplicate rows: {duplicates}")
    
    # Statistical overview for numerical columns
    numerical_cols = df.select_dtypes(include=['float64', 'int64']).columns
    if len(numerical_cols) > 0:
        print(f"\n📈 Statistical Summary for Numerical Columns:")
        display(df[numerical_cols].describe())
    
    # Rating analysis
    if 'vote_average' in df.columns:
        print(f"\n⭐ Movie Rating Analysis:")
        print(f"   Highest Rating: {df['vote_average'].max()}/10")
        print(f"   Lowest Rating: {df['vote_average'].min()}/10")
        print(f"   Average Rating: {df['vote_average'].mean():.2f}/10")
        print(f"   Movies with 8+ rating: {(df['vote_average'] >= 8).sum():,}")
    
    # Popularity insights
    if 'popularity' in df.columns:
        print(f"\n📈 Popularity Analysis:")
        print(f"   Most Popular Score: {df['popularity'].max():.1f}")
        print(f"   Least Popular Score: {df['popularity'].min():.1f}")
        print(f"   Average Popularity: {df['popularity'].mean():.1f}")
    
    # Release date analysis
    if 'release_date' in df.columns:
        # Convert to datetime for analysis
        df['release_date'] = pd.to_datetime(df['release_date'], errors='coerce')
        print(f"\n📅 Release Date Analysis:")
        print(f"   Earliest Movie: {df['release_date'].min().strftime('%Y-%m-%d') if df['release_date'].min() else 'N/A'}")
        print(f"   Latest Movie: {df['release_date'].max().strftime('%Y-%m-%d') if df['release_date'].max() else 'N/A'}")
        
    # Display sample of the data
    print(f"\n🔍 Sample Movies from Dataset:")
    display(df.head(10))
    
except Exception as e:
    print(f"❌ Analysis error: {e}")
    print(f"📊 Basic info - Dataset shape: {df.shape if 'df' in locals() else 'DataFrame not found'}")
````
