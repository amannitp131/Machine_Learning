## 🌐 3. Reading CSV from URLs

### 🎯 Learning Objective:
Learn to fetch and load CSV data directly from web URLs for remote data access.

### 📋 Key Concepts:
- **HTTP Requests**: Making web requests to fetch data
- **User-Agent Headers**: Identifying your request to avoid blocking
- **StringIO**: Converting text data to file-like objects
- **Remote Data Access**: Accessing datasets hosted online

### 💡 Why This Matters:
Many datasets are hosted online, and direct URL reading enables:
- ✅ **Real-time Data**: Access to live, updated datasets
- ✅ **Collaboration**: Shared access to common datasets
- ✅ **Reproducibility**: Consistent data source across environments

### 🔧 Implementation:

```python
# Import required libraries for web requests
import requests
from io import StringIO

# Define the URL and headers
url = "https://raw.githubusercontent.com/cs109/2014_data/master/countries.csv"
headers = {
    "User-Agent": "Mozilla/5.0 (Macintosh; Intel Mac OS X 10.14; rv:66.0) Gecko/20100101 Firefox/66.0"
}


# Make the HTTP request
req = requests.get(url, headers=headers)

# Convert the response text to a file-like object
data = StringIO(req.text)

# Read the CSV data
df_countries = pd.read_csv(data)

# Display the DataFrame
df_countries.head()
```

## 🌍 10. Encoding Parameter

### 🎯 Learning Objective:
Handle different text encodings to prevent character display issues and reading errors.

### 📋 Key Concepts:
- **Text Encoding**: How characters are stored in bytes
- **UTF-8**: Universal encoding (most common)
- **Latin-1**: Western European characters
- **ASCII**: Basic English characters only
- **Encoding Errors**: Character corruption prevention

### 💡 Common Encodings:
- ✅ **utf-8**: Default, handles most languages
- ✅ **latin-1**: European languages, legacy files
- ✅ **ascii**: Basic English only
- ✅ **cp1252**: Windows encoding

### 🚨 Signs of Encoding Issues:
- Strange characters (�, ?, weird symbols)
- UnicodeDecodeError exceptions
- Missing or corrupted text

### 🔧 Implementation:

```python
# Using encoding parameter to handle files with special characters
# latin-1 encoding is often needed for files with European characters

print("🌍 Reading file with latin-1 encoding...")

try:
    # Read CSV with specific encoding to handle special characters
    df_zomato = pd.read_csv('zomato.csv', encoding='latin-1')
    
    print(f"✅ File loaded successfully with latin-1 encoding!")
    print(f"📊 Dataset Shape: {df_zomato.shape}")
    print(f"📋 Columns: {df_zomato.columns.tolist()}")
    print("\n🔍 First 5 rows:")
    
    # Display the DataFrame
    df_zomato.head()
    
except UnicodeDecodeError as e:
    print(f"❌ Encoding error: {e}")
    print("💡 Try different encoding like 'utf-8', 'cp1252', or 'ascii'")
except FileNotFoundError:
    print("📁 File 'zomato.csv' not found in current directory")
```

## 🚨 11. Error Handling - Skip Bad Lines

### 🎯 Learning Objective:
Learn to handle corrupted or malformed rows in CSV files gracefully.

### 📋 Key Concepts:
- **Data Quality Issues**: Malformed rows, inconsistent columns
- **Error Recovery**: Continue processing despite bad data
- **Data Validation**: Identify and handle problematic records
- **Robust Loading**: Prevent crashes from data issues

### 💡 Common Data Problems:
- ⚠️ **Extra Columns**: Rows with more fields than headers
- ⚠️ **Missing Columns**: Rows with fewer fields
- ⚠️ **Delimiter Issues**: Inconsistent separators
- ⚠️ **Quote Problems**: Unclosed quotes or escaping issues

### 🔧 Error Handling Options:
- **on_bad_lines='error'**: Stop and raise error (default)
- **on_bad_lines='warn'**: Show warning and skip
- **on_bad_lines='skip'**: Silently skip bad lines

### 🔧 Implementation:

```python
# Handling corrupted data with error handling
# Using on_bad_lines parameter to skip problematic rows

print("🚨 Reading file with error handling...")

try:
    # Read CSV with bad line handling (modern pandas syntax)
    # Note: error_bad_lines is deprecated, use on_bad_lines instead
    df_books = pd.read_csv(
        'BX-Books.csv', 
        sep=';', 
        encoding="latin-1",
        on_bad_lines='skip'  # Skip bad lines instead of crashing
    )
    
    print(f"✅ File loaded with error handling!")
    print(f"📊 Dataset Shape: {df_books.shape}")
    print("⚠️ Any malformed rows were automatically skipped")
    print("\n🔍 First 5 rows:")
    
    # Display the DataFrame
    df_books.head()
    
except Exception as e:
    print(f"❌ Error loading file: {e}")
    print("💡 File may not exist or have severe formatting issues")
```

## ❌ 15. Custom Missing Values (na_values)

### 🎯 Learning Objective:
Define custom values to be treated as missing/null during data loading.

### 📋 Key Concepts:
- **Missing Data Identification**: Custom null value recognition
- **Data Quality**: Proper handling of various null representations
- **Standardization**: Consistent missing value treatment
- **Domain Knowledge**: Business-specific null indicators

### 💡 Common Missing Value Indicators:
- ✅ **Standard**: NaN, None, null, NULL
- ✅ **Custom**: "N/A", "Unknown", "Missing"
- ✅ **Placeholders**: -999, 0, "TBD"
- ✅ **Domain-specific**: "Not Applicable", "Refused"

### 🔧 Implementation:

```python
# Define custom values to be treated as missing/null
# This is useful when your data uses non-standard null indicators

print("❌ Loading CSV with custom missing value indicators...")

# Define custom values that should be treated as NaN
custom_na_values = ['Male', 'Unknown', 'Not Specified', 'N/A']

# Read CSV treating custom values as missing
df_custom_na = pd.read_csv('aug_train.csv', na_values=custom_na_values)


# Check the effect on a specific column (if 'gender' exists)
if 'gender' in df_custom_na.columns:
    print(f"\n🔍 Gender column after custom NA treatment:")
    print(f"Value counts:\n{df_custom_na['gender'].value_counts(dropna=False)}")
    print(f"Missing values: {df_custom_na['gender'].isna().sum()}")
else:
    print("Gender column not found - showing general missing value info")
    
# Show overall missing value summary
print(f"\n📊 Missing values per column:")
missing_summary = df_custom_na.isnull().sum()
print(missing_summary[missing_summary > 0])  # Show only columns with missing values
```

## 🔄 16. Chunked Reading for Large Datasets

### 🎯 Learning Objective:
Process large datasets efficiently using chunk-based reading for memory management.

### 📋 Key Concepts:
- **Memory Management**: Handle files larger than available RAM
- **Chunk Processing**: Read data in manageable pieces
- **Iterative Analysis**: Process data batch by batch
- **Scalability**: Handle any file size regardless of memory

### 💡 When to Use Chunked Reading:
- ✅ **Large Files**: Files too big for memory
- ✅ **Limited RAM**: Working with memory constraints
- ✅ **Streaming Analysis**: Real-time data processing
- ✅ **Batch Operations**: Process data in segments

### 🔧 Implementation:

```python
# Create a chunk reader for large dataset processing
# chunksize parameter returns an iterator instead of loading all data at once

print("🔄 Setting up chunked reading for large dataset...")

# Create a chunk reader with specified chunk size
chunk_size = 5000
chunk_reader = pd.read_csv('aug_train.csv', chunksize=chunk_size)

print(f"✅ Chunk reader created!")
print(f"📦 Chunk size: {chunk_size} rows")
print(f"🔄 Reader type: {type(chunk_reader)}")
print("\n💡 This creates an iterator - no data loaded yet!")
print("Use the iterator in a loop to process chunks one by one.")
```

```python
# Process chunks iteratively for memory-efficient analysis
# This demonstrates how to work with chunked data

print("🔄 Processing chunks iteratively...")

# Recreate chunk reader (iterators can only be used once)
chunk_reader = pd.read_csv('aug_train.csv', chunksize=5000)

chunk_count = 0
total_rows = 0

# Process each chunk
for chunk in chunk_reader:
    chunk_count += 1
    total_rows += len(chunk)
    
    print(f"📦 Chunk {chunk_count}: Shape {chunk.shape}")
    
    # Example analysis on each chunk
    if 'gender' in chunk.columns:
        print(f"   Gender distribution: {chunk['gender'].value_counts().to_dict()}")
    
    # Limit output for demonstration
    if chunk_count >= 3:  # Process only first 3 chunks for demo
        print("   ... (stopping after 3 chunks for demonstration)")
        break

print(f"\n✅ Processed {chunk_count} chunks")
print(f"📊 Total rows processed: {total_rows}")
print("💡 This approach allows processing datasets of any size!")
```
