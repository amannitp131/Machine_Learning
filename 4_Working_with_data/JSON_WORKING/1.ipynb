## 📁 3. Loading Local JSON Files

### 🎯 Learning Objective:
Master reading JSON files from your local file system.

### 📋 Key Concepts:
- **Local File Access**: Reading JSON from disk
- **Automatic Parsing**: Pandas handles JSON structure conversion
- **DataFrame Creation**: JSON objects become rows and columns
- **Nested Data**: Handling complex JSON hierarchies

### 💡 JSON File Structure:
```json
{
  "data": [
    {"id": 1, "name": "Alice", "age": 25},
    {"id": 2, "name": "Bob", "age": 30}
  ]
}
```

### 🔧 Implementation:

````python
# Load local JSON file into pandas DataFrame
# JSON files are automatically parsed and converted to tabular format

print("📁 Loading local JSON file...")

try:
    # Read the local JSON file
    df_json = pd.read_json('train.json')
    
    # Display the DataFrame
    display(df_json.head())
    
except FileNotFoundError:
    print("❌ File 'train.json' not found in current directory")
    print("💡 Make sure the JSON file exists in the working directory")
except ValueError as e:
    print(f"❌ JSON parsing error: {e}")
    print("💡 Check if the JSON file is properly formatted")
````

## 🌐 4. Loading JSON from APIs

### 🎯 Learning Objective:
Access real-time data from web APIs using JSON format.

### 📋 Key Concepts:
- **API Integration**: Direct data access from web services
- **Real-time Data**: Live, updated information
- **HTTP Requests**: Automatic web data fetching
- **Currency Exchange**: Practical API example

### 💡 API Benefits:
- ✅ **Live Data**: Always current information
- ✅ **No Storage**: No need to download files
- ✅ **Standardized**: JSON is web standard
- ✅ **Automation**: Easy to integrate in workflows

### 🔧 Implementation:

````python
# Load JSON data directly from a web API
# This example uses a currency exchange rate API

print("🌐 Fetching live data from Exchange Rate API...")

try:
    # Read JSON data from API endpoint
    api_url = 'https://api.exchangerate-api.com/v4/latest/INR'
    df_exchange = pd.read_json(api_url)

    # Display the DataFrame
    display(df_exchange)
    
    # Additional analysis if 'rates' column exists
    if 'rates' in df_exchange.columns:
        print(f"\n🔍 Available currencies: {len(df_exchange['rates'].iloc[0])} currencies")
        print(f"💱 Sample rates from INR:")
        rates = df_exchange['rates'].iloc[0]
        for currency, rate in list(rates.items())[:5]:  # Show first 5 rates
            print(f"   1 INR = {rate:.4f} {currency}")
    
except Exception as e:
    print(f"❌ Error loading API data: {e}")
    print("💡 Check internet connection or API availability")
````

## 🔗 7. Database Connection Setup

### 🎯 Learning Objective:
Establish a secure connection to MySQL database.

### 📋 Key Concepts:
- **Connection Parameters**: Host, user, password, database
- **Authentication**: Secure database access
- **Connection Object**: Interface for database operations
- **Error Handling**: Managing connection failures

### 💡 Connection Parameters:
- **host**: Database server location ('localhost' for local)
- **user**: Database username ('root' is default admin)
- **password**: User authentication (empty for development)
- **database**: Specific database name ('world' sample database)

### 🔧 Implementation:

````python
# Establish connection to MySQL database
# Using sample 'world' database with local MySQL server

print("🔗 Establishing database connection...")

try:
    # Create database connection
    conn = mysql.connector.connect(
        host='localhost',      # Database server (local machine)
        user='root',          # Database username
        password='',          # Database password (empty for development)
        database='world'      # Target database name
    )
    
    print("✅ Database connection established successfully!")
    print(f"🏢 Connected to: {conn.database}")
    print(f"🌐 Server: {conn.server_host}:{conn.server_port}")
    print(f"👤 User: {conn.user}")
    print("🔒 Connection is secure and ready for queries")
    
except mysql.connector.Error as e:
    print(f"❌ Database connection failed: {e}")
    print("💡 Check if MySQL server is running and credentials are correct")
    print("💡 Make sure 'world' database exists or change database name")
````

## 📊 8. SQL Query Execution

### 🎯 Learning Objective:
Execute SQL queries and load results directly into pandas DataFrames.

### 📋 Key Concepts:
- **SQL Queries**: Database query language for data retrieval
- **pandas.read_sql_query()**: Bridge between SQL and DataFrame
- **Table Access**: Querying specific database tables
- **Data Integration**: Seamless SQL-to-pandas workflow

### 💡 SQL Query Benefits:
- ✅ **Powerful Filtering**: Complex WHERE clauses
- ✅ **Data Joining**: Multiple table relationships
- ✅ **Aggregation**: GROUP BY and statistical functions
- ✅ **Performance**: Database-optimized operations

### 🔧 Implementation:

````python
# Execute SQL query and load results into pandas DataFrame
# Query the 'countrylanguage' table from the world database

print("📊 Executing SQL query...")

try:
    # Execute SQL query and create DataFrame
    sql_query = "SELECT * FROM countrylanguage"
    df_languages = pd.read_sql_query(sql_query, conn)
    
    print(f"✅ SQL query executed successfully!")
    print(f"📊 Query Results Shape: {df_languages.shape}")
    print(f"📋 Columns: {df_languages.columns.tolist()}")
    print(f"🗄️ Table: countrylanguage")
    print(f"📝 Query: {sql_query}")
    
    # Display basic statistics
    print(f"\n📈 Dataset Statistics:")
    print(f"   Total records: {len(df_languages)}")
    if 'CountryCode' in df_languages.columns:
        print(f"   Unique countries: {df_languages['CountryCode'].nunique()}")
    if 'Language' in df_languages.columns:
        print(f"   Unique languages: {df_languages['Language'].nunique()}")
    
    print("\n🔍 First 5 rows:")
    
except Exception as e:
    print(f"❌ SQL query failed: {e}")
    print("💡 Check if table 'countrylanguage' exists in the database")
    print("💡 Verify database connection is still active")
````
