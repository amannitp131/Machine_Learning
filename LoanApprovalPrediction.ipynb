```python
df=pd.read_csv("loan_approval_dataset.csv")
df.head()
```

```python
df.drop(columns=["loan_id"],axis=1,inplace=True)
```

```python
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python

df.columns=[col.strip() for col in df.columns]
df['education']=[val.strip() for val in df['education']]
df['self_employed']=[val.strip() for val in df['self_employed']]
```

```python
df.columns
```

```python
X=df.drop(columns=["loan_status"],axis=1)
Y=df['loan_status']
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.25,random_state=42)
```

```python
X_train.shape,X_test.shape,Y_train.shape,Y_test.shape
```

```python
Y_train.value_counts()
```

```python
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
Y_train=le.fit_transform(Y_train)
Y_test=le.transform(Y_test)
```

```python
Y_train,Y_test
```

```python
cat_col=[feature for feature in X_train.columns if X_train[feature].dtype=='O']
num_col=[feature for feature in X_train.columns if X_train[feature].dtype!='O']
```

```python
X_train['education'].value_counts()
X_train['self_employed'].value_counts()
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder,StandardScaler

cat_transformer=OneHotEncoder()
num_transformer=StandardScaler()

preprocessor=ColumnTransformer(transformers=[('cat',cat_transformer,cat_col),('num',num_transformer,num_col)])

X_train=preprocessor.fit_transform(X_train)
X_test=preprocessor.transform(X_test)

```

```python
pd.DataFrame(X_train)
```

```python
# use Smote for handling imbalanced dataset

from imblearn.over_sampling import SMOTE
smote=SMOTE()
X_train,Y_train=smote.fit_resample(X_train,Y_train)

Y_train = pd.Series(Y_train)
Y_train.value_counts()
```

```python





# Sklearn Models
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, AdaBoostClassifier, GradientBoostingClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.svm import SVC


from xgboost import XGBClassifier

# Dictionary of models
models = {
    "LogisticRegression": LogisticRegression(),
    "DecisionTreeClassifier": DecisionTreeClassifier(),
    "RandomForestClassifier": RandomForestClassifier(),
    "KNeighborsClassifier": KNeighborsClassifier(),
    "GaussianNB": GaussianNB(),

    "XGBClassifier": XGBClassifier(use_label_encoder=False, eval_metric='logloss'),
    "AdaBoostClassifier": AdaBoostClassifier(),
    "GradientBoostingClassifier": GradientBoostingClassifier()
}

```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score,classification_report,precision_score,recall_score,roc_auc_score,f1_score


for name, model in models.items():
    model.fit(X_train, Y_train)  # Train the model

    # Make predictions
    Y_train_pred = model.predict(X_train)
    Y_test_pred = model.predict(X_test)

    # Compute training performance metrics
    train_accuracy = accuracy_score(Y_train, Y_train_pred)
    train_f1 = f1_score(Y_train, Y_train_pred)
    train_precision = precision_score(Y_train, Y_train_pred)
    train_recall = recall_score(Y_train, Y_train_pred)
    train_roc_auc = roc_auc_score(Y_train, model.predict_proba(X_train)[:, 1])

    # Compute test performance metrics
    test_accuracy = accuracy_score(Y_test, Y_test_pred)
    test_f1 = f1_score(Y_test, Y_test_pred)
    test_precision = precision_score(Y_test, Y_test_pred)
    test_recall = recall_score(Y_test, Y_test_pred)
    test_roc_auc = roc_auc_score(Y_test, model.predict_proba(X_test)[:, 1])

    # Print results in a formatted manner
    print(f"\nPerformance Metrics for {name}:")
    print("=" * 40)
    print(f"Training Accuracy: {train_accuracy:.4f}")
    print(f"Training F1 Score: {train_f1:.4f}")
    print(f"Training Precision: {train_precision:.4f}")
    print(f"Training Recall: {train_recall:.4f}")
    print(f"Training ROC AUC Score: {train_roc_auc:.4f}")
    print("-" * 40)
    print(f"Test Accuracy: {test_accuracy:.4f}")
    print(f"Test F1 Score: {test_f1:.4f}")
    print(f"Test Precision: {test_precision:.4f}")
    print(f"Test Recall: {test_recall:.4f}")
    print(f"Test ROC AUC Score: {test_roc_auc:.4f}")
    print("=" * 40)
```

```python
## select top 3 models ->GradientBoostingClassifier,XGBClassifier,RandomForestClassifier

gb_params={
    "loss":['log_loss', 'exponential'],
    "learning_rate":[0.01,0.05,0.1,0.2,0.5],
    "n_estimators":[100,200,500,1000],
    "criterion":['friedman_mse', 'squared_error']

}

rf_params={
    "n_estimators":[100,200,500,1000],
    "criterion":['gini', 'entropy', 'log_loss'],
    "max_depth":[5,8,15,10,None],
    "min_samples_split":[2,8,15,10],
    "max_features":["sqrt","log2",None]
}

xgb_params={
    'learning_rate': [0.01, 0.05, 0.1, 0.2, 0.5],
    'n_estimators': [100,200,300,150]

}
```

```python
randomcv_models=[
    ("GradientBoostingClassifier",GradientBoostingClassifier(),gb_params),
    ("RandomForestClassifier",RandomForestClassifier(),rf_params),
    ("XGBClassifier",XGBClassifier(),xgb_params)
]
```

```python
from sklearn.model_selection import RandomizedSearchCV

for name, model, params in randomcv_models:
    print(f"\nPerforming RandomizedSearchCV for {name}...")

    # Define RandomizedSearchCV
    random_search = RandomizedSearchCV(
        estimator=model,
        param_distributions=params,
        n_iter=10,  # Number of iterations, can be adjusted
        scoring='accuracy',  # Metric for evaluation
        cv=5,  # 5-fold cross-validation
        n_jobs=-1,  # Use all processors for efficiency
        random_state=42
    )

    # Fit RandomizedSearchCV on training data
    random_search.fit(X_train, Y_train)

    # Retrieve best parameters
    best_params = random_search.best_params_

    # Print results in a readable format
    print("\n" + "=" * 40)
    print(f"Best Hyperparameters for {name}:")
    print("-" * 40)
    for param, value in best_params.items():
        print(f"{param}: {value}")
    print("=" * 40)
```

```python
## create our final model

models={
    "GradientBoostingClassifier":GradientBoostingClassifier(criterion='friedman_mse', learning_rate=0.1, loss='log_loss', n_estimators=500),
    "RandomForestClassifier":RandomForestClassifier(criterion='gini', max_depth=None, max_features='sqrt', min_samples_split=8, n_estimators=500),
    "XGBClassifier":XGBClassifier(learning_rate=0.5, n_estimators=200)
}

```

```python
for name, model in models.items():
    model.fit(X_train, Y_train)  # Train the model

    # Make predictions
    Y_train_pred = model.predict(X_train)
    Y_test_pred = model.predict(X_test)

    # Compute training performance metrics
    train_accuracy = accuracy_score(Y_train, Y_train_pred)
    train_f1 = f1_score(Y_train, Y_train_pred)
    train_precision = precision_score(Y_train, Y_train_pred)
    train_recall = recall_score(Y_train, Y_train_pred)
    train_roc_auc = roc_auc_score(Y_train, model.predict_proba(X_train)[:, 1])

    # Compute test performance metrics
    test_accuracy = accuracy_score(Y_test, Y_test_pred)
    test_f1 = f1_score(Y_test, Y_test_pred)
    test_precision = precision_score(Y_test, Y_test_pred)
    test_recall = recall_score(Y_test, Y_test_pred)
    test_roc_auc = roc_auc_score(Y_test, model.predict_proba(X_test)[:, 1])

    # Print results in a formatted manner
    print(f"\nPerformance Metrics for {name}:")
    print("=" * 40)
    print(f"Training Accuracy: {train_accuracy:.4f}")
    print(f"Training F1 Score: {train_f1:.4f}")
    print(f"Training Precision: {train_precision:.4f}")
    print(f"Training Recall: {train_recall:.4f}")
    print(f"Training ROC AUC Score: {train_roc_auc:.4f}")
    print("-" * 40)
    print(f"Test Accuracy: {test_accuracy:.4f}")
    print(f"Test F1 Score: {test_f1:.4f}")
    print(f"Test Precision: {test_precision:.4f}")
    print(f"Test Recall: {test_recall:.4f}")
    print(f"Test ROC AUC Score: {test_roc_auc:.4f}")
    print("=" * 40)
```
