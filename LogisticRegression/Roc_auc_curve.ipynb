```python
from sklearn.model_selection import train_test_split
from sklearn.datasets import make_classification
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import roc_curve
from sklearn.metrics import roc_auc_score
from matplotlib import pyplot

```

```python
x,y=make_classification(n_samples=1000,n_classes=2,random_state=42)
```

```python
## do train test split 
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.25,random_state=43)
```

```python
## cretae a dummy model having default op as 0

dummy_model_prob=[0 for _ in range(len(y_test))]
dummy_model_prob
```

```python
model=LogisticRegression()
model.fit(x_train,y_train)
```

```python
model_prob=model.predict_proba(x_test)
model_prob
```

```python
## lets focus only on positive probability
model_prob=model_prob[:,1]
model_prob
```

```python
dummy_model_auc=roc_auc_score(y_test,dummy_model_prob)
model_auc=roc_auc_score(y_test,model_prob)
print(dummy_model_auc)
print(model_auc)
```

```python
dummy_fpr, dummy_tpr, dummy_thresholds = roc_curve(y_test, dummy_model_prob)
model_fpr, model_tpr, model_thresholds = roc_curve(y_test, model_prob)
```

```python
pyplot.plot(dummy_fpr,dummy_tpr,linestyle="--",label="dummy")
pyplot.plot(model_fpr,model_tpr,marker='.',label="logistic")


for i in range(0, len(model_thresholds), max(1, len(model_thresholds)//10)):  # annotate every ~10th point
    pyplot.annotate(f"{model_thresholds[i]:.2f}", 
                 (model_fpr[i], model_tpr[i]), 
                 textcoords="offset points", 
                 xytext=(0, -15), 
                 ha='center',
                 fontsize=8, color='red')


pyplot.xlabel("false positive rate ")
pyplot.ylabel("true positive rate")

pyplot.legend()
pyplot.show()
```
