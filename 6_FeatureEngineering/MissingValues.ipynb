# Imputation Technique

## mean value Imputation

```python
sns.histplot(df['age'],kde=True)
```

```python
## since the distribution of data is some what like the gausssian distribution.so we can replace missing values with the mean 
## of the column 


df['age']=df['age'].fillna(df['age'].mean())
df.isnull().sum()
```

## Median Imputation

```python
## median imputation is used when our data has a outliers 
sns.histplot(df['deck'],kde=True)
```

```python
df['age']=df['age'].fillna(df['age'].median())
```

```python
df.head()
```

## Mode Imputation 

```python
## it is used for categorical variable 

df['embarked'].unique()
```

```python
df['embarked']=df['embarked'].mode()[0]
```
