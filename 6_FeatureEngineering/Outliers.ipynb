# 5 Number Summary And Box Plot

```python
lst_marks=[90,98,70,67,89,45,56,67,32,12,56,87,86,90,92,91,88,77]
```

```python
## calculate minimum,Q1,median,Q3,maximum

import numpy as np
minimum,Q1,median,Q3,maximum=np.quantile(lst_marks,[0.0,0.25,0.50,0.75,1.0])
```

```python
minimum,Q1,median,Q3,maximum
```

```python
## calculate IQR

IQR=Q3-Q1
print(IQR)
```

```python
##calculate lower and upper fence 

lower_fence =Q1-1.5*(IQR)
upper_fence=Q3+1.5*(IQR)
```

```python
outliers=[]
for marks in lst_marks:
    if(marks<lower_fence or marks>upper_fence):
        outliers.append(marks)
```

```python
outliers
```

```python
## visulaize outliers 


import seaborn as sns 

sns.boxplot(lst_marks)
```
