# One Hot Encoding -> used for columns having more than two categories 

```python
df=pd.DataFrame({
    'colors':["red","green","blue","red","blue"]
})

df.head()
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
encoder=OneHotEncoder()
encoded=encoder.fit_transform(df[['colors']]).toarray()
```

```python
final_df=pd.DataFrame(encoded,columns=encoder.get_feature_names_out())
```

```python
final_df
```

```python
pd.concat([final_df,df],axis=1)
```

```python
## practical with dataset

import seaborn as sns 

df=sns.load_dataset('tips')
```

```python
df.head()
```

```python
from sklearn.preprocessing import OneHotEncoder
encoder=OneHotEncoder()
```

```python
encoded_sex=encoder.fit_transform(df[['sex']]).toarray()

encoded_sex_df=pd.DataFrame(encoded_sex,columns=encoder.get_feature_names_out())
encoded_sex_df.head()
```

```python
## now drop original sex column 

df.drop(columns=["sex"],axis=1,inplace=True)
```

```python
pd.concat([df,encoded_sex_df],axis=1)
```

# Ordinal Encoding

```python
# ## example->
# High school->1
# college->2
# graduate->3
# post_graduate->4
```

```python
from sklearn.preprocessing import OrdinalEncoder
```

```python
df=pd.DataFrame({
    'size':["small","medium","large","medium","small","large"]
})
```

```python
## create an instance of ordinal encoder

encoder=OrdinalEncoder(categories=[["small","medium","large"]])

```

```python
encoded_size=encoder.fit_transform(df[['size']])
```

```python
encoded_size_df=pd.DataFrame(encoded_size,columns=["encoded_size"])
```

```python
final_df=pd.concat([df,encoded_size_df],axis=1)
```

```python
final_df
```

# Target Guided Ordinal Encoding

```python
## THEORY:-
## in the target guided ordinal encoding we replace each category in the categorical variable with the numerical value based on mean or median of the target variable for each category
```

```python
import pandas as pd
df=pd.DataFrame(
    {
        'city':["new york","london","paris","tokyo","new york","paris"],
        'price':[200,150,300,250,180,320]
    }
)
```

```python
city_mean=df.groupby('city')["price"].mean().to_dict()
```

```python
df['city_Encoded']=df['city'].map(city_mean)
```
