# Feature Engineering Notes

## What is Feature Engineering?
Feature engineering is the process of transforming raw data into meaningful features that improve the performance of machine learning models. It involves creating, modifying, or selecting features to enhance model accuracy and efficiency.

---

## 1. Classifying Feature Engineering
- **Feature Creation:** Generating new features from existing data, such as extracting year from a date or creating interaction terms.
- **Feature Transformation:** Modifying features to better represent the underlying problem, e.g., applying log transformation to skewed data.
- **Feature Selection:** Choosing the most relevant features for modeling to reduce overfitting and improve interpretability.

---

## 2. Missing Values Imputation
- **Mean/Median/Mode Imputation:** Replace missing values with the mean (for continuous), median (for skewed), or mode (for categorical) values.
- **Forward/Backward Fill:** Use previous or next values to fill gaps, often used in time series data.
- **Model-based Imputation:** Predict missing values using other features (e.g., KNN imputer, regression imputer).
- **Dropping:** Remove rows or columns with too many missing values if imputation is not feasible.

---

## 3. Handling Categorical Values
- **Label Encoding:** Assigns each category a unique integer. Suitable for ordinal data.
- **One-Hot Encoding:** Creates binary columns for each category. Used for nominal data.
- **Ordinal Encoding:** Encodes categories with an inherent order (e.g., low, medium, high).
- **Frequency/Count Encoding:** Replace categories with their frequency or count in the dataset.

---

## 4. Outlier Detection
- **Statistical Methods:**
  - *Z-score:* Points with a Z-score above a threshold (e.g., 3) are considered outliers.
  - *IQR Method:* Points outside 1.5*IQR above Q3 or below Q1 are outliers.
- **Visualization:**
  - *Box plots:* Show distribution and highlight outliers.
  - *Scatter plots:* Visualize relationships and spot anomalies.
- **Model-based:**
  - *Isolation Forest, DBSCAN:* Algorithms that can detect outliers in high-dimensional data.

---

## 5. Feature Scaling
- **Standardization:** Scales features to have zero mean and unit variance. Useful for algorithms like SVM, KNN.
- **Normalization:** Scales features to a fixed range (e.g., 0 to 1). Useful for neural networks.
- **Robust Scaling:** Uses median and IQR for scaling, less sensitive to outliers.

---

## 6. Feature Construction
- **Polynomial Features:** Create interaction terms or higher-order features (e.g., x1*x2, x1^2).
- **Aggregation:** Summarize groups of features (e.g., mean, sum, min, max).
- **Domain Knowledge:** Use expertise to create meaningful features, such as ratios or flags.
- **Date/Time Features:** Extract year, month, day, hour, or time since an event.

---

## 7. Feature Selection
- **Filter Methods:**
  - *Correlation:* Remove highly correlated features.
  - *Chi-square test:* For categorical variables.
- **Wrapper Methods:**
  - *Recursive Feature Elimination (RFE):* Iteratively remove least important features.
- **Embedded Methods:**
  - *Lasso Regression:* Shrinks less important feature coefficients to zero.
  - *Tree-based Methods:* Use feature importance from decision trees or random forests.

---

## 8. Feature Extraction
- **PCA (Principal Component Analysis):** Reduces dimensionality by combining features into principal components.
- **t-SNE, UMAP:** Non-linear dimensionality reduction techniques for visualization.
- **Autoencoders:** Neural network-based feature extraction for complex data.

---

**Summary:**
Feature engineering is a crucial step in the machine learning pipeline. It improves model performance, interpretability, and robustness by carefully preparing and selecting the right features for your data and problem.
