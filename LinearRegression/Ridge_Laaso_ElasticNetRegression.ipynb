## Feature selection

```python
x=df_copy.drop(columns=["FWI"],axis=1)
y=df_copy["FWI"]
```

```python
x.head()
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
corr=x_train.corr()
plt.figure(figsize=(12,10))
sns.heatmap(corr,annot=True)
```

```python


def get_highly_correlated_features(dataset: pd.DataFrame, threshold: float):
    correlation_matrix = dataset.corr().abs()
    correlated_features = []

    for i in range(len(correlation_matrix.columns)):
        for j in range(i + 1, len(correlation_matrix.columns)):
            if correlation_matrix.iloc[i, j] > threshold:
                feature_1 = correlation_matrix.columns[i]
                
                correlated_features.append(feature_1)

    return correlated_features

```

```python
col=get_highly_correlated_features(x_train,0.90)
```

```python
## drop features when coreelation is more than 0.85
x_train.drop(columns=col,axis=1,inplace=True)
```

```python
x_test.drop(columns=col,axis=1,inplace=True)
```

```python
## scaling

from sklearn.preprocessing import StandardScaler
scalar=StandardScaler()
x_train_scaled=scalar.fit_transform(x_train)
x_test_scaled=scalar.fit_transform(x_test)
```

```python
## box plot to understand the effect of outlier
plt.figure(figsize=(10,6))
sns.boxplot(x_train)
plt.title("before scaling")
```

```python
## box plot to understand the effect of outlier
plt.figure(figsize=(10,6))
sns.boxplot(x_train_scaled)
plt.title("after scaling")
```

## Model Training

### linear regression

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, r2_score

linreg = LinearRegression()
linreg.fit(x_train_scaled, y_train)
y_pred = linreg.predict(x_test_scaled)

mae = mean_absolute_error(y_test, y_pred)  
score = r2_score(y_test, y_pred)           

print("mean absolute error is : ", mae)
print("r2 score is : ", score)
```

```python
## debugging 

print(x_train_scaled.shape)
print(y_train.shape)
```

## Lasso Regression

```python
from sklearn.linear_model import Lasso
from sklearn.metrics import mean_absolute_error, r2_score

lasso = Lasso()
lasso.fit(x_train_scaled, y_train)
y_pred = lasso.predict(x_test_scaled)

mae = mean_absolute_error(y_test, y_pred)  
score = r2_score(y_test, y_pred)           

print("mean absolute error is : ", mae)
print("r2 score is : ", score)
```

## Ridge Regression 

```python
from sklearn.linear_model import Ridge
from sklearn.metrics import mean_absolute_error, r2_score

ridge = Ridge()
ridge.fit(x_train_scaled, y_train)
y_pred = ridge.predict(x_test_scaled)

mae = mean_absolute_error(y_test, y_pred)  
score = r2_score(y_test, y_pred)           

print("mean absolute error is : ", mae)
print("r2 score is : ", score)
```

## Elastic net regression

```python
from sklearn.linear_model import ElasticNet
from sklearn.metrics import mean_absolute_error, r2_score

elastic = ElasticNet()
elastic.fit(x_train_scaled, y_train)
y_pred = elastic.predict(x_test_scaled)

mae = mean_absolute_error(y_test, y_pred)  
score = r2_score(y_test, y_pred)           

print("mean absolute error is : ", mae)
print("r2 score is : ", score)
```

# HyperParameter Tuning 

```python
from sklearn.linear_model import Ridge
from sklearn.model_selection import GridSearchCV
```

```python
ridge = Ridge()

# Try different alpha values (log scale is common)
params = {
    'alpha': [0.0001, 0.001, 0.01, 0.1, 1, 10, 100, 1000]
}
```

```python
grid = GridSearchCV(ridge, params, cv=5, scoring='r2')  # or scoring='neg_mean_squared_error'
grid.fit(x_train_scaled, y_train)
```

```python
print("Best alpha:", grid.best_params_)
print("Best R² score on training data:", grid.best_score_)
```

```python
## try on test data
best_ridge = grid.best_estimator_
y_pred = best_ridge.predict(x_test_scaled)

from sklearn.metrics import mean_absolute_error, r2_score

print("Test MAE:", mean_absolute_error(y_test, y_pred))
print("Test R² score:", r2_score(y_test, y_pred))
```
