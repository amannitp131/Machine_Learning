## Handle Date of Journey 

```python
## handle date of journey

## our model will not understand the data so we break the date into three parts day ,month and year

df['date']=df['Date_of_Journey'].str.split('/').str[0]
df['month']=df['Date_of_Journey'].str.split('/').str[1]
df['year']=df['Date_of_Journey'].str.split('/').str[2]
```

```python
df.head()
```

```python
df.info()
```

```python
df['date']=df['date'].astype('int')
df['month']=df['month'].astype('int')
df['year']=df['year'].astype('int')
```

```python
df.info()
```

```python
## drop date of journey column
df.drop(columns=["Date_of_Journey"],axis=1,inplace=True)
```

```python
df.head()
```

## Handle Arrival Time

```python
# df['Arrival_Time'].str.split(':').str[1].str.split(" ").str[0]

df['arrival_hour']=df['Arrival_Time'].str.split(":").str[0]
df["arrival_minute"]=df['Arrival_Time'].str.split(":").str[1].str.split(" ").str[0]
```

```python
df.head()
```

```python
df.info()
```

```python
df['arrival_hour']=df['arrival_hour'].astype('int')
df['arrival_minute']=df['arrival_minute'].astype('int')
```

```python
df.info()
```

```python
## drop arrival_time 
df.drop(columns=["Arrival_Time"],axis=1,inplace=True)
```

```python
df.head()
```

## Handle Departure Time

```python
df["departure_hour"]=df["Dep_Time"].str.split(":").str[0]
df["departure_minute"]=df["Dep_Time"].str.split(":").str[1]


```

```python
df.head(2)
```

```python
df.info()
```

## Handle Total_Stops

```python
df['Total_Stops'].unique()
```

```python
df['Total_Stops'].isnull().sum()
```

```python
df.isnull().sum()
```

```python
## since we have only 2 rows with null value drop these 

df.dropna(inplace=True)
```

```python
df.isnull().sum()
```

```python
## use ordinal encoder to encode the Total_stops

from sklearn.preprocessing import OrdinalEncoder
encoder=OrdinalEncoder(categories=[["non-stop","1 stop","2 stops","3 stops","4 stops"]])
encoded_total_stops=encoder.fit_transform(df[["Total_Stops"]])
```

```python
encoded_total_stops
```

```python
new_df=pd.DataFrame(encoded_total_stops,columns=["encoded_total_stops"])
```

```python
new_df
```

```python
df=pd.concat([df,new_df],axis=1)
```

```python
df.head()
```

```python
df.drop(columns=["Total_Stops"],axis=1,inplace=True)
```

```python
df.head()
```

```python
df.info()
```

## Handle Duration

```python
df["Duration_hour"] = 0
df["Duration_min"] = 0

df.loc[df["Duration"].str.contains("h", na=False), "Duration_hour"] = (
    df["Duration"].str.extract(r'(\d+)h')[0].fillna(0).astype(int)
)

df.loc[df["Duration"].str.contains("m", na=False), "Duration_min"] = (
    df["Duration"].str.extract(r'(\d+)m')[0].fillna(0).astype(int)
)
```

```python
df.head()
```

```python
df.info()
```

```python
## drop duration column
df.drop(columns=["Duration"],axis=1,inplace =True)
```

```python
df.head(2)
```

```python
df["Source"].unique()
```
