# 📊 Seaborn Visualization Guide

Seaborn is a powerful Python library for creating informative and attractive statistical graphics. This guide covers essential plots and their concise implementations.

## 🔹 Scatter Plot
Used to visualize the relationship between two continuous variables. Example: Total Bill vs Tip.
## 🔹 Line Plot
Shows trends or changes over time or categories. Example: Size vs Total Bill.
## 🔹 Bar Plot
Compares categorical data with aggregated values. Example: Day vs Total Bill.
## 🔹 Box Plot
Summarizes data distribution and detects outliers. Example: Day vs Total Bill.
## 🔹 Violin Plot
Combines box plot and KDE for richer insights. Example: Day vs Total Bill.
## 🔹 Histogram
Visualizes the distribution of a single variable. Example: Total Bill.
## 🔹 Pair Plot
Explores pairwise relationships in a dataset. Example: Tips Dataset.
## 🔹 Heatmap
Displays correlations or matrix data. Example: Correlation Matrix.

```python
import seaborn as sns
```

```python
## basic plotting using seaborn 
tips=sns.load_dataset("tips")
tips
```

```python
import matplotlib.pyplot as plt 
```

```python
sns.scatterplot(x='total_bill' ,y='tip',data=tips)
plt.title("scatter plot of total bill vs tips")
```

```python
tips.dtypes
```

```python
## line plot 
sns.lineplot(x="size",y="total_bill",data=tips)
plt.title("line plot of size of member vs total bill")
plt.show()
```

```python
## categorical variables plot 
sns.barplot(x="day",y="total_bill",data=tips)
plt.title("bar graph of day vs total bill")
plt.show()
```

```python
## box plot 
sns.boxplot(x="day",y="total_bill",data=tips)
plt.title("day vs total_bill box plot")
plt.show()
```

```python
## violin plot 
sns.violinplot(x="day",y="total_bill",data=tips)
plt.title("day vs total_bill violin plot")
plt.show()
```

```python
## histograms 
sns.histplot(tips["total_bill"],bins=10,kde=True)
```

```python
sns.kdeplot(tips["total_bill"])
```

```python
## pair plot
sns.pairplot(tips)
```

```python
## heat map
corr=tips[["total_bill","tip","size"]].corr()
corr
## note x->axis=tip y->axis=totalbill value=0.67 means if we increase 1 unit of tip then total bill is increased by 0.67 unit 
```

```python
sns.heatmap(corr,annot=True,cmap="viridis")
```
